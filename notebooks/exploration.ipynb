# DocBank Exploration
Quick notebook for checking parsed samples and label distribution before training.

```python
from pathlib import Path
from collections import Counter

from src.data.preprocessing import load_samples

data_dir = Path('./data/docbank/DocBank_500K_txt')
samples = load_samples(data_dir=data_dir, max_samples=200, ignored_labels=['figure', 'table', 'equation', 'caption', 'figure_caption', 'table_caption'])
len(samples)
```

```python
counter = Counter()
for s in samples:
    counter.update(s.labels)
counter.most_common(20)
```
